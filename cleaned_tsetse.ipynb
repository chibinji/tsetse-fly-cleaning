# Tsetse Fly Survey 2013: Data Cleaning

**Steps:** load the data, check size and columns, handle missing values, check duplicates, fix data types and names, then save a cleaned CSV.

### Import libraries
pandas for data handling, numpy for numeric work

```python
import pandas as pd
import numpy as np
```

### Load the data
Read the 2013 tsetse fly survey file into a DataFrame and display the first 5 rows

```python
data = pd.read_csv("DDDAC_tsetse_2013.csv")
data.head()
```

### Check the size
The dataset starts with 4,497 rows and 10 columns.

```python
data.shape
```

### List the columns

```python
data.columns
```

### Check for missing values
Shows the percentage of nulls per column. Only Latitude_ and Longitude_ have missing values (about 2.5%)

```python
data.isnull().sum()/len(data) *100
```

### Drop rows with missing coordinates
Remove the rows where Latitude_ is null (the missing longitudes are in the same rows). Other columns are unaffected. This leaves 4,384 rows.

```python
data.dropna(subset=['Latitude_'], inplace=True)
```

### Confirm no missing values remain

```python
data.isnull().sum()/len(data) * 100
```

### Check for duplicate rows

```python
data.duplicated().sum()
```

### Check for unique values


```python
for col in data.columns:
    print(col, data[col].unique())
```

### Fix the stray S in Result
One row has S instead of 0/1. Its Count is 0, so it is treated as a typo for 0.

```python
print(data['Result'].value_counts())
data['Result'] = data['Result'].replace('S', '0').astype(int)
```

### Convert `Date` to a real datetime

```python
data['Date'] = pd.to_datetime(data['Date'], format='%d/%m/%Y')
data['Date'].dtype
```

### Rename the coordinate columns

```python
data = data.rename(columns={'Latitude_': 'Latitude', 'Longitude_': 'Longitude'})
data.dtypes
```

### Save the cleaned data


```python
data.to_csv('cleaned_tsetse_data.csv', index=False)
```
